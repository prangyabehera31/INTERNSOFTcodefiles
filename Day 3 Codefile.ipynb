```python
import math
def is_prime(num):
    if num % 2 == 0 and num > 2:
        return False
    for i in range(3, int(math.sqrt(num)) +1,2): #(start,stop,step size)
        if num % i == 0:
            return False
    return True       
```

```python
is_prime(11)
```

```python
is_prime(25)
```

```python
# generate fnction for the cube of numbers  (power of 3)
def cube(n): 
    for num in range(n):
        yield num **3
```

```python
for x in cube(10):
    print(x)
```

```python
def fibon(n):
    a=1
    b=1
    for i in range(n):
        yield a
        a,b = b,a+b
```

```python
for num in fibon(10):
    print(num)
```

```python
def fibon(n):
    a=1
    b=1
    output = []
    for i in range(n):
        output.append(a)
        a,b=b,a++b
    return output    
    
```

```python
fibon(5)
```

```python
def fahrenheit(T):
    return ((float(9)/5)*T + 32)
def celsius(T):
    return (float(5)/9)*(T-32)
temp = [0,22.5,40,100]    
```

```python
F_temps = list(map(fahrenheit,temp))
F_temps               
```

```python
list(map(celsius, F_temps))
```

```python
list(map(lambda x: (5.0/9)*(x - 32), F_temps))
```
