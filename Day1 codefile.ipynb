```python
#Program to find the sum of all numbers store in list

#List of numbers
numbers = [6, 5, 3, 8, 4, 2, 5, 4, 9]

#variable to store the sum
sum = 0

#iterate over the list
for val in numbers:
    sum = sum+val
print("The sum is",sum)    
sum
```

```python
digits=[0, 1, 5]
for i in digits:
    print(i)
else:
    print("No items left")
```

```python
#Program to add natural
#numbers upto
#sum = 1+2+3+...+n

# To take input from user,
n=int(input("Enter n:"))

#initialize sum and counter
sum=0
i=1

while i <= n:
      sum = sum+1
      i= i + 1  #update counter
print("The sum is",sum)
```
